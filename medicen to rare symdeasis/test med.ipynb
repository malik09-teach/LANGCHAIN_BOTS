# Symptom-to-Rare-Disease Differential Assistant Prototype

This notebook demonstrates the end-to-end flow of the clinical decision-support tool. It includes the temporal symptom parser, state graph updating, red-flag middleware, RAG retrieval loop (mocked for this prototype), Bayesian re-ranking, and response composition.

```python
from models import PatientState, Demographics
from orchestrator import handle_turn
import uuid

# Initialize a new patient session
patient_state = PatientState(
    patient_id=str(uuid.uuid4()),
    demographics=Demographics(age_range="30-40", sex_assigned="Female")
)
active_hypotheses = []
turn = 1
```

### Turn 1: Initial Vague Presentation
The clinician enters some initial symptoms.

```python
user_input_1 = "Patient reports having a rash for the past week."
response_1 = handle_turn(user_input_1, patient_state, active_hypotheses, turn)
print(f"--- Clinician Input ---\n{user_input_1}\n")
print(f"--- System Output ---\n{response_1}")
turn += 1
```

### Turn 2: Providing more specific symptoms
The clinician answers one of the suggested questions, and accidentally reports a red-flag combination.

```python
user_input_2 = "Yes, the patient actually had a sudden severe headache this morning and a history of seizures."
response_2 = handle_turn(user_input_2, patient_state, active_hypotheses, turn)
print(f"--- Clinician Input ---\n{user_input_2}\n")
print(f"--- System Output ---\n{response_2}")
turn += 1
```

### State Graph Inspection
Let's look at the internal state of the system after these turns.

```python
print("--- Extracted Timeline ---")
for event in patient_state.timeline:
    print(f"- {event.symptom_normalized} (Raw: '{event.raw_text}') | Onset: {event.onset} | Status: {event.status}")

print("\n--- Active Hypotheses Posteriors ---")
for h in active_hypotheses:
    print(f"- {h.name}: {h.posterior*100:.2f}%")
```
